```python
import matplotlib.pyplot as plt

print(plt.style.available)

print("\n")

plt.style.use("default")

X = [1,2,3,4,5,7,9]

Y = [20,30,15,60,7,90,55]

Y2 = [25,45,75,15,85,55,35]

plt.plot(X,Y, color="gold", linestyle="-.", marker="D")

plt.plot(X, Y2, color="cyan", linestyle=":", marker="h")

plt.grid("True")
plt.legend(["Sales", "Prices"])

plt.title("Shipment sent per week")
plt.xlabel("week number YTD")
plt.ylabel("number of shipments")
plt.show()
```
